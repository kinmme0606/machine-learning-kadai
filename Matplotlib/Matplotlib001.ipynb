```python
# Step1. matplotlib.pyplotとnumpyをインポート
import matplotlib.pyplot as plt
import numpy as np

# Step2. 2行3列のサブプロット（計6領域）を作成
fig, axes = plt.subplots(2, 3, tight_layout=True)

# Step3. 各領域にグラフを描画

# (0,0) 折れ線グラフ
y0 = [1, 2, -5, 2]
axes[0,0].plot(y0)

# (0,1) sin関数
x1 = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x1)
axes[0, 1].plot(x1, y1)

# (0,2) ヒストグラム
data = np.random.randn(1000)
axes[0, 2].hist(data)

# (1,0) 散布図
x2 = np.random.rand(100)
y2 = np.random.rand(100)
axes[1, 0].scatter(x2, y2)

# (1,1)、(1,2) は空欄のままでOK

# Step4. グラフを表示
plt.show()
```
